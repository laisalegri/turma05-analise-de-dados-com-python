## 2. Dado, Informação, Conhecimento e Tipo

| Conceito | Definição |
|---|---|
| Dado | Registro bruto, sem contexto |
| Tipo | A natureza desse dado (número, texto, data...) |
| Informação | Dado organizado, com contexto |
| Conhecimento | Informação aplicada a uma decisão |

### 🔹 Exemplo 1 — Temperatura corporal

Veja como isso aparece em código: o **dado** vira uma **variável**, e o `print()` mostra a **informação** formatada.

> 🔍 **Onde fica a fronteira entre Dado e Informação?** Se eu só dissesse "`152`", sem dizer mais nada, isso ainda seria só **dado** — um número solto, sem contexto. Só vira **informação** quando dizemos o que ele representa (veja o Exemplo 2 abaixo: "vendemos 152 unidades hoje, 20% acima da média"). O número sozinho não muda — o que muda é o contexto que colocamos em volta dele.
>
> 💡 **O padrão por trás do "Conhecimento":** quase toda decisão segue a mesma receita: *informação + um critério de comparação (faixa, meta, limite) = conhecimento*. Guarde esse padrão — ele vai se repetir o curso inteiro.

````python
# DADO: um número guardado em uma variável
temperatura = 37

# INFORMAÇÃO: o dado organizado, com contexto, mostrado na tela
print(f"A temperatura do paciente é {temperatura}°C")

# CONHECIMENTO: a decisão, comparando a informação com um critério/faixa conhecida
# a faixa de temperatura corporal considerada normal é 36°C a 37,5°C
print("37°C está dentro da faixa normal (36°C–37,5°C) — não preciso agir.")
````

## 3. As Etapas da Análise de Dados

```
1. COLETA        → reunir os dados
2. LIMPEZA       → corrigir erros, remover duplicatas e nulos
3. ANÁLISE       → buscar padrões e causas
4. VISUALIZAÇÃO  → transformar números em gráficos
5. COMUNICAÇÃO   → apresentar o resultado e a recomendação
```

Essas etapas vão se repetir em **todas as semanas seguintes** do curso.

#### Cenário B — E-commerce com muitas devoluções

**Problema**

Uma loja online identificou que 1 em cada 5 pedidos está sendo devolvido nos últimos 2 meses. O squad deve aplicar as **5 etapas da análise de dados** para descobrir **por que** o índice de devoluções é tão alto e propor uma ação concreta que a loja possa implementar.

**Solução**

O dado crítico não é o volume de devoluções — é o **motivo**. Sem ele, sabemos que algo está errado, mas não onde agir. O caminho: coletar pedidos + devoluções com os motivos registrados, padronizar variações do mesmo motivo na Limpeza ("produto errado" = "item incorreto" = o mesmo problema), analisar por categoria e por motivo, visualizar os produtos mais devolvidos, e comunicar a causa raiz com ação específica. A descoberta-chave é que "produto diferente da descrição" aponta para problema interno (fotos e textos das SKUs), não para qualidade do produto em si — essa distinção muda completamente a recomendação.

````python
cenario_do_squad = "E-commerce com muitas devoluções — 1 em cada 5 pedidos retornado nos últimos 2 meses"
print("Cenário:", cenario_do_squad)

print("1. Coleta:", "dados do sistema de pedidos (produto, valor, canal de venda, data) e dados de devoluções (motivo registrado, data, produto devolvido)")
print("2. Limpeza:", "remover devoluções duplicadas (mesmo pedido com mais de um registro), padronizar motivos escritos de formas diferentes ('produto errado' vs 'item incorreto'), tratar datas faltantes")
print("3. Análise:", "calcular taxa de devolução por produto, por categoria e por motivo; identificar se há concentração em uma marca ou fornecedor específico; comparar os 2 meses com o histórico anterior")
print("4. Visualização:", "gráfico de barras com os 10 produtos mais devolvidos; gráfico de pizza com os motivos mais frequentes; gráfico de linha com a evolução mensal da taxa de devolução")
print("5. Comunicação:", "os produtos da categoria Eletrônicos têm taxa de devolução 3x acima da média — o motivo mais frequente é 'produto diferente da descrição'. Recomendamos revisar fotos e textos dessas SKUs antes de novas campanhas")

pergunta_para_outros_squads = "Quando a análise mostra que 60% das devoluções são de um só produto, como vocês distinguiriam se o problema é a qualidade do produto, a descrição falsa ou a expectativa incorreta do cliente?"
print("Pergunta para os outros squads:", pergunta_para_outros_squads)
````

#### Cenário C — Hospital com tempo de espera crescente

**Problema**

Pacientes estão reclamando que o tempo de espera no pronto-atendimento aumentou. O squad deve aplicar as **5 etapas da análise de dados** para identificar **quando e por que** o tempo de espera cresceu e propor uma intervenção que o hospital possa executar.

**Solução**

O sintoma é o tempo de espera, mas a causa pode ser volume de pacientes, número de médicos, tipo de urgência ou sazonalidade (gripe). O dado que desvenda este caso é cruzar o tempo médio de espera por horário com a **escala de plantão**: quando o turno da tarde passou de 2 para 1 médico, o tempo triplicou — e isso aparece no dado, sem precisar de testemunho. Atenção especial na Limpeza: os dados faltantes provavelmente se concentram no horário de pico (o sistema travou exatamente quando mais precisava registrar). Na Comunicação, "falta médico no turno da tarde" é politicamente sensível — o analista apresenta o dado ("com 1 médico, espera = 90min; com 2, espera = 35min"), não o julgamento.

````python
cenario_do_squad = "Hospital com tempo de espera crescente no pronto-atendimento"
print("Cenário:", cenario_do_squad)

print("1. Coleta:", "registros de entrada de pacientes (hora de chegada, hora da triagem, hora do atendimento, hora da alta); escala de plantão dos médicos; classificação de urgência; volume de pacientes por dia e por hora")
print("2. Limpeza:", "tratar registros sem horário de atendimento (sistema travou durante pico?); remover entradas duplicadas do mesmo paciente; padronizar classificação de urgência registrada de formas diferentes por plantonistas distintos")
print("3. Análise:", "calcular tempo médio de espera por hora do dia, por dia da semana e por nível de urgência; verificar se o aumento coincide com mudança no número de médicos por turno, com sazonalidade (gripe, feriados) ou com tipo de queixa")
print("4. Visualização:", "gráfico de linha com tempo médio de espera por hora do dia; heatmap com dia da semana × hora; gráfico de barras comparando tempo de espera por turno antes e depois do aumento")
print("5. Comunicação:", "o tempo médio de espera entre 13h e 16h subiu de 35 para 90 minutos — no mesmo período, o turno da tarde passou de 2 para 1 médico. Recomendamos reforçar o quadro neste horário ou criar um fluxo de triagem acelerada para casos leves")

pergunta_para_outros_squads = "Como vocês lidariam com dados de saúde de pacientes durante a análise? O que muda quando os dados são sensíveis pela LGPD?"
print("Pergunta para os outros squads:", pergunta_para_outros_squads)
````

#### Cenário D — Indústria com quebras frequentes de máquina

**Problema**

Uma fábrica percebeu que uma máquina específica está quebrando com frequência crescente. O squad deve aplicar as **5 etapas da análise de dados** para identificar **o que mudou** para causar as quebras e propor uma correção antes da próxima parada não planejada.

**Solução**

O desafio central é separar **quebra real** de parada planejada — o sistema de manutenção mistura os dois, e essa distinção exige conversa com o técnico (lição: Limpeza nunca é só técnica, é negócio). O dado que desvenda o caso é o **ponto de inflexão**: ao plotar as quebras por mês, o salto de 4x após março aponta para uma mudança naquele período — e o log revela troca de fornecedor de rolamentos. Leituras de sensor impossíveis (ex.: 900°C num sensor calibrado para 0–200°C) são falha do sensor, não da máquina — tratá-las como dado inválido na Limpeza evita uma conclusão falsa na Análise.

````python
cenario_do_squad = "Indústria com quebras frequentes da máquina de produção"
print("Cenário:", cenario_do_squad)

print("1. Coleta:", "logs de manutenção (data, tipo de intervenção, peça trocada, técnico responsável); dados de operação (horas ligada, carga de produção, tipo de produto fabricado no momento da quebra); leituras de sensores (temperatura, vibração, pressão); notas fiscais de peças de reposição")
print("2. Limpeza:", "remover registros de paradas planejadas (manutenção preventiva não é quebra); corrigir datas inseridas retroativamente; tratar leituras de sensor impossíveis (pico de 900°C em sensor de 0–200°C indica falha do sensor, não da máquina)")
print("3. Análise:", "contar quebras por mês antes e depois de uma data-chave; verificar se as quebras se concentram em um turno, um operador ou um tipo de produto fabricado; identificar se houve troca de peça ou de fornecedor perto do início do problema")
print("4. Visualização:", "gráfico de linha com número de quebras por mês; linha do tempo sobrepondo datas de manutenção e quebras; gráfico de barras por turno e por operador")
print("5. Comunicação:", "as quebras aumentaram 4 vezes após março, quando houve troca de fornecedor de rolamentos — as peças do novo fornecedor têm vida útil estimada 60% menor. Recomendamos retornar ao fornecedor anterior ou exigir laudo técnico de qualidade antes de usar o novo lote")

pergunta_para_outros_squads = "Os sensores geram leituras a cada segundo. Com meses de dados, são milhões de linhas. Como vocês decidiriam qual granularidade usar na análise — segundo, minuto ou hora?"
print("Pergunta para os outros squads:", pergunta_para_outros_squads)
````

#### Cenário E — Prefeitura com reclamações de buracos nas ruas

**Problema**

Moradores estão reclamando nas redes sociais sobre buracos nas ruas. O squad deve aplicar as **5 etapas da análise de dados** para transformar reclamações dispersas em informação acionável: onde estão os buracos, quais bairros são mais afetados e se o tempo de conserto está dentro do aceitável.

**Solução**

Este é o cenário de Coleta mais complexa: dados de redes sociais são texto livre, não estruturado. A Limpeza tem triplo desafio — endereços informais ("perto da padaria do João"), posts duplicados sobre o mesmo buraco e posts irônicos. O dado de contexto que transforma o diagnóstico é a **precipitação** (chuva): cruzar volume de reclamações com a curva de chuva revela que os picos ocorrem depois de eventos de chuva forte em bairros com asfalto antigo — um achado que muda a priorização do investimento. A visualização geográfica (mapa de calor por bairro) comunica muito mais do que um gráfico de barras aqui.

````python
cenario_do_squad = "Prefeitura com aumento de reclamações sobre buracos nas ruas"
print("Cenário:", cenario_do_squad)

print("1. Coleta:", "posts de redes sociais com menção a buracos (Twitter/X, Instagram); reclamações registradas no canal 156; ordens de serviço de tapa-buracos emitidas e concluídas; dados de precipitação (chuva gera buracos e pode explicar picos de reclamação)")
print("2. Limpeza:", "remover posts duplicados do mesmo morador sobre o mesmo buraco; extrair localização de texto livre ('perto do mercado da esquina' não é endereço válido); filtrar posts irônicos ou não relacionados a buracos reais; padronizar bairros escritos de formas diferentes")
print("3. Análise:", "mapear concentração geográfica das reclamações por bairro; calcular tempo médio entre reclamação e conserto; identificar se picos de reclamação coincidem com períodos de chuva intensa; comparar volume de reclamações com ordens de serviço concluídas")
print("4. Visualização:", "mapa de calor com distribuição geográfica dos buracos; gráfico de linha com volume de reclamações por semana sobreposto à curva de precipitação; gráfico de barras com tempo médio de resposta por bairro")
print("5. Comunicação:", "os bairros X, Y e Z concentram 68% das reclamações, com tempo médio de conserto de 47 dias — 3x acima da meta de 15 dias. Recomendamos priorizar essas regiões e revisar o fluxo de abertura de ordens de serviço para reduzir o tempo de resposta")

pergunta_para_outros_squads = "Como vocês tratariam reclamações onde o morador não informa o endereço exato, só diz 'tem um buraco perto da padaria do João'? Isso é um problema de Coleta ou de Limpeza?"
print("Pergunta para os outros squads:", pergunta_para_outros_squads)
````
