```python
import pandas as pd
from data_profiling import ProfileReport

df = pd.read_csv('vendas.csv', parse_dates=['data_venda'])
```

```python
import sweetviz as sv

report = sv.analyze(df)
report.show_html('relatorio_sweetviz.html')
```

```python
primeiro = df[df['data_venda'].dt.month <= 6]
segundo = df[df['data_venda'].dt.month > 6]
comparacao = sv.compare([primeiro, '1º Semestre'], [segundo, '2º Semestre'])
comparacao.show_html('relatorio_comparacao.html')
```
